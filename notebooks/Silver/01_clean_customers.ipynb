```python
%skip 

# Imports 
import pyspark.sql.functions as F
from datetime import datetime
import uuid
from pyspark.sql.window import Window 
from delta.tables import DeltaTable 

from src.utils.logger import get_logger
from src.utils.audit import write_audit_log
from src.orchestration.batch import get_batch_id 

start_time = datetime.now()
error_message = None
rows_read = 0
rows_written = 0 
batch_id = get_batch_id(spark)

try: 
    logger = get_logger("silver.clean_customers", log_to_file=True)
    
    logger.info(f"Starting Silver pipeline at {start_time}")

    # Configurations 
    CATALOG = "catalog_project1"
    BRONZE_SCHEMA = "bronze"
    SILVER_SCHEMA = "silver"

    SOURCE_TABLE = f"{CATALOG}.{BRONZE_SCHEMA}.customers"
    TARGET_TABLE = f"{CATALOG}.{SILVER_SCHEMA}.clean_customers"

    # Check if it exists 
    if not spark.catalog.tableExists(SOURCE_TABLE): 
        raise ValueError(f"Source table {SOURCE_TABLE} does not exist")

    email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
    pipeline_id = str(uuid.uuid4())

    # Read bronze 
    cust_bronze = spark.read.table(SOURCE_TABLE)
    logger.info(f"Reading {SOURCE_TABLE} table")

    # Validate input 
    cust_bronze.printSchema() 
    # cust_bronze.show(10, truncate=False) 

    # Check for empty source table 
    if cust_bronze.isEmpty(): 
        raise Exception(f"No records found in the {SOURCE_TABLE}") 
    
    rows_read = cust_bronze.count()
    logger.info(f"Read {rows_read} records from {SOURCE_TABLE}") 

    # Required column validation 
    required_columns = ["customer_id", "name", "email", "city", "state", "signup_date", "created_date", "modified_date",    "_ingestion_job_id"] 

    actual_columns = set(cust_bronze.columns)

    missing_columns = set(required_columns) - actual_columns 

    if missing_columns: 
        raise ValueError(f"Missing columns {missing_columns} in the {SOURCE_TABLE}") 
    
    logger.info(f"All required columns exists in the {SOURCE_TABLE}") 

    # Select the columns 
    cust_clean = cust_bronze.select(
        "customer_id", 
        "name", 
        "email", 
        "city", 
        "state", 
        "signup_date", 
        "created_date", 
        "modified_date", 
        "_ingestion_job_id" 
    )

    # Standardize the data 
    cust_clean = cust_clean.select(
        F.trim("customer_id").alias("customer_id"),
        F.initcap(F.trim("name")).alias("name"),
        F.lower(F.trim("email")).alias("email"), 
        F.initcap(F.trim("city")).alias("city"),
        F.initcap(F.trim("state")).alias("state"), 
        F.trim("signup_date").alias("signup_date"), 
        F.trim("created_date").alias("created_date"), 
        F.trim("modified_date").alias("modified_date"), 
        F.trim("_ingestion_job_id").alias("_ingestion_job_id") 
    ) 

    logger.info(f"Data standardization completed for the {SOURCE_TABLE}")

    # cust_clean.printSchema() 

    # Cast data types 
    cust_clean = cust_clean.select(
        F.col("customer_id").cast("integer"), 
        F.col("name").cast("string"), 
        F.col("email").cast("string"), 
        F.col("city").cast("string"), 
        F.col("state").cast("string"), 
        F.col("signup_date").cast("timestamp"), 
        F.col("created_date").cast("timestamp"), 
        F.col("modified_date").cast("timestamp"), 
        F.col("_ingestion_job_id").cast("string") 
    ) 

    logger.info(f"Data type casting completed for the {SOURCE_TABLE}")

    # Check for null customer ids 
    if cust_clean.filter(F.col("customer_id").isNull()).count() > 0: 
        raise ValueError(f"Null customer ids found in the {SOURCE_TABLE}")

    # Check for Nulls, and Validate  
    invalid_records = cust_clean.filter(
        (F.col("customer_id").isNull()) |
        (F.col("customer_id") < 0) |
        (F.col("name").isNull()) |
        (F.col("name").rlike(r"^[a-zA-Z ]+$") == False) |
        (F.col("email").isNull()) |
        (F.col("email").rlike(email_pattern) == False) |
        (F.col("city").isNull()) |
        (F.col("city").rlike(r"^[a-zA-Z ]+$") == False) |
        (F.col("state").isNull()) |
        (F.col("state").rlike(r"^[a-zA-Z ]+$") == False) |
        (F.col("signup_date").isNull()) |
        (F.col("signup_date") < datetime(1900, 1, 1)) |
        (F.col("created_date").isNull()) |
        (F.col("created_date") < datetime(1900, 1, 1)) |
        (F.col("modified_date").isNull()) |
        (F.col("modified_date") < datetime(1900, 1, 1))
    ).count() 

    if invalid_records > 0: 
        print(f"Invalid records found in the {SOURCE_TABLE}") 
        raise Exception(f"Invalid records found in the {SOURCE_TABLE}") 

    logger.info(f"Data validation completed for the {SOURCE_TABLE}")

    # Check for duplicates 
    duplicates = cust_clean.groupBy("customer_id").count().filter("count > 1").count() 

    # Deduplication logic to keep only latest records based on modified_date - SCD1 
    window_spec = (
        Window
        .partitionBy("customer_id")
        .orderBy(F.desc("modified_date"))
    ) 

    cust_latest = (
        cust_clean
        .withColumn("_row_number", F.row_number().over(window_spec))
        .filter(F.col("_row_number") == 1)
        .drop("_row_number")
    ) 

    logger.info(f"Deduplication completed for the {SOURCE_TABLE}")

    # Add metadata 
    cust_latest = cust_latest.select(
        F.col("*"), 
        # F.col("_metadata.file_path").alias("_source_file"),
        F.lit(SOURCE_TABLE).alias("_source_table"), 
        F.lit(pipeline_id).alias("_ingestion_pipeline_id"), 
        F.current_timestamp().alias("ingestion_timestamp"), 
        F.current_timestamp().alias("updation_timestamp") 
    ) 

    logger.info(f"Writing Metadata for the {SOURCE_TABLE}")

    # More Data quality checks 
    """
    Completeness(NOT NULL), Uniqueness, Validity, Consistency, Accuracy, Timeliness, Referential integrity 
    """ 

    # display(cust_latest)

    # Write: Overwrite => Merge 
    # Check if table exists
    if not spark.catalog.tableExists(TARGET_TABLE):
        # Initial write if table doesn't exist
        cust_latest.write.mode("overwrite")\
                        .format("delta")\
                        .saveAsTable(TARGET_TABLE)
        logger.info(f"Created new table {TARGET_TABLE}")
    else:
        # Merge logic for existing table
        target = DeltaTable.forName(spark, TARGET_TABLE) 
        (
            target.alias("t")\
              .merge(
                  source = cust_latest.alias("s"), 
                  condition = F.expr("t.customer_id = s.customer_id")
              )\
            #   .whenMatchedUpdateAll()\
              .whenMatchedUpdate(
                  condition = F.expr("t.modified_date <> s.modified_date"), 
                  set = {
                      "name": "s.name", 
                      "email": "s.email", 
                      "city": "s.city", 
                      "state": "s.state", 
                      "signup_date": "s.signup_date", 
                      "created_date": "s.created_date", 
                      "modified_date": "s.modified_date", 
                      "_ingestion_job_id": "s._ingestion_job_id",
                      "ingestion_timestamp": "s.ingestion_timestamp", 
                      "updation_timestamp": "current_timestamp()", 
                      "_source_table": "s._source_table", 
                      "_ingestion_pipeline_id": "s._ingestion_pipeline_id" 
                  }
              )\
              .whenNotMatchedInsertAll()\
              .execute()
        )

    logger.info(f"Data written to the {TARGET_TABLE}")

    # Verify the output 
    silver_df = spark.read.table(TARGET_TABLE) 
    silver_df.printSchema()
    # Row count 
    rows_written = silver_df.count() 
    # Uniqueness 
    duplicates = (
        silver_df
        .groupBy("customer_id").count()
        .filter("count > 1").count()
    ) 
    if duplicates > 0: 
        print(f"Duplicate records found in the {TARGET_TABLE}") 
        raise Exception(f"Duplicate records found in the {TARGET_TABLE}") 
    # Check for Nulls, and Validate 
    nulls = silver_df.filter(
        (F.col("customer_id").isNull()) |
        (F.col("customer_id") < 0) |
        (F.col("name").isNull()) |
        (F.col("name").rlike(r"^[a-zA-Z ]+$") == False) |
        (F.col("email").isNull()) |
        (F.col("email").rlike(email_pattern) == False) |
        (F.col("city").isNull()) |
        (F.col("city").rlike(r"^[a-zA-Z ]+$") == False) |
        (F.col("state").isNull()) |
        (F.col("state").rlike(r"^[a-zA-Z ]+$") == False) |
        (F.col("signup_date").isNull()) |
        (F.col("signup_date") < datetime(1900, 1, 1)) |
        (F.col("created_date").isNull()) |
        (F.col("created_date") < datetime(1900, 1, 1)) |
        (F.col("modified_date").isNull()) |
        (F.col("modified_date") < datetime(1900, 1, 1))
    ).count() 
    if nulls > 0: 
        print(f"Invalid records found in the {TARGET_TABLE}") 
        raise Exception(f"Invalid records found in the {TARGET_TABLE}") 
    
    logger.info(f"Data validation for {TARGET_TABLE} completed")
    status = "SUCCESS"

except Exception as e: 
    status = "FAILURE"
    error_message = str(e) 
    # print(f"Silver {TARGET_TABLE} pipeline failed: {error_message}")
    logger.exception(f"Silver {TARGET_TABLE} pipeline failed")
    raise 

finally: 
    end_time = datetime.now() 
    logger.info(f"Silver pipeline for {TARGET_TABLE} completed successfully")
    try: 
        write_audit_log(pipeline = "silver_pipeline", table = "clean_customers", start_time = start_time, end_time = end_time, rows_read = rows_read, rows_written = rows_written, status = status, error_message = error_message, CATALOG = CATALOG, AUDIT_SCHEMA = "audit", AUDIT_TABLENAME = "pipeline_run_audit", batch_id = batch_id, watermark_column = "modified_date", watermark_value = None) 
        logger.info("Audit written successfully")
    except Exception as e: 
        logger.exception(f"Error writing audit log: {e}")
```
