Note: The root cause is clear: the `src` directory exists as a workspace directory, not as a Python package directory accessible via standard Python imports on Serverless compute.

So, created setup.py in root folder to create a proper python project that can be imported anywhere. 

Install project as a python package using `%pip install -e <path>` 

Then restart the kernal using `dbutils.library.restartPython()` to use updated packages. 

### How setup.py + find_packages() Works

`find_packages()` scans your project directory and finds every folder that contains an `__init__.py` file. In your case, it found:

- `src/`
- `src/ingestion/`
- `src/utils/`
- etc. 

What `%pip install -e .` does:

When you run `%pip install -e /path/to/project` (the `-e` means "editable mode"):

1. Creates a link in Python's site-packages directory pointing to your project folder
2. Registers the packages found by `find_packages()` as importable modules
3. Adds metadata so Python knows where to find these modules

After this, `import src` works from anywhere because Python knows `src` is a registered package.



The restart clears everything — the editable install doesn't persist after restartPython(). And on Serverless, without the restart, the installation doesn't fully activate.



### Approach used 
The `sys.path` approach: 

How this works:

1. `sys.path` is Python's list of directories where it searches for modules
2. By adding your project root, Python now looks there when you `import src`
3. Since `src/` has an `__init__.py`, Python treats it as a package
4. No installation needed — it's just path manipulation
