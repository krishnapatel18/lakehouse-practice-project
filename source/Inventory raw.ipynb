```python
from pyspark.sql import Row
from pyspark.sql.types import * 
from pyspark.sql.functions import to_timestamp 

data = [
    Row(inventory_id=1, product_id=501, warehouse="Ahmedabad_WH", stock=25, last_updated="01-03-2025", created_date="01-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=2, product_id=502, warehouse="Mumbai_WH", stock=100, last_updated="01-03-2025", created_date="03-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=3, product_id=503, warehouse="Delhi_WH", stock=80, last_updated="01-03-2025", created_date="05-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=4, product_id=504, warehouse="Bengaluru_WH", stock=20, last_updated="01-03-2025", created_date="10-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=5, product_id=505, warehouse="Ahmedabad_WH", stock=30, last_updated="01-03-2025", created_date="15-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=6, product_id=506, warehouse="Mumbai_WH", stock=500, last_updated="01-03-2025", created_date="18-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=7, product_id=507, warehouse="Delhi_WH", stock=120, last_updated="01-03-2025", created_date="20-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=8, product_id=508, warehouse="Bengaluru_WH", stock=70, last_updated="01-03-2025", created_date="25-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=9, product_id=509, warehouse="Ahmedabad_WH", stock=90, last_updated="01-03-2025", created_date="28-12-2024", modified_date="01-03-2025"),
    Row(inventory_id=10, product_id=510, warehouse="Mumbai_WH", stock=40, last_updated="01-03-2025", created_date="30-12-2024", modified_date="01-03-2025")
]

schema = StructType([
    StructField("inventory_id", IntegerType(), False),
    StructField("product_id", IntegerType(), False),
    StructField("warehouse", StringType(), True),
    StructField("stock", IntegerType(), True),
    StructField("last_updated", StringType(), True),
    StructField("created_date", StringType(), True),
    StructField("modified_date", StringType(), True)
])

inventory_raw = spark.createDataFrame(data, schema) 

inventory_raw = inventory_raw \
    .withColumn("last_updated", to_timestamp("last_updated", "dd-MM-yyyy")) \
    .withColumn("created_date", to_timestamp("created_date", "dd-MM-yyyy")) \
    .withColumn("modified_date", to_timestamp("modified_date", "dd-MM-yyyy"))

display(inventory_raw)
```

```python
# spark.sql("DROP TABLE IF EXISTS catalog_project1.source1.inventory_raw")
inventory_raw.write.mode("append")\
                  .format("delta")\
                  .option("mergeSchema", "true")\
                  .saveAsTable("catalog_project1.source1.inventory_raw")
```
