```python
from pyspark.sql import Row
from pyspark.sql.types import *
from pyspark.sql.functions import to_timestamp 

# Incremental data with 1 update and 1 new record
data = [
    Row(supplier_id=111, supplier_name="IT TECH      ", contact_name="        Zhao Lusi", contact_email="zhaolusi@ittech.com", country="China", city="Beijing", created_date="01-12-2024", modified_date="01-12-2024"), 
    Row(supplier_id=110, supplier_name="Home Needs", contact_name="Sunita Rao", contact_email="sunita@homeneeds.com", country="India", city="Delhi", created_date="19-11-2024", modified_date="01-12-2024")
]

schema = StructType([
    StructField("supplier_id", IntegerType(), False),
    StructField("supplier_name", StringType(), True),
    StructField("contact_name", StringType(), True),
    StructField("contact_email", StringType(), True),
    StructField("country", StringType(), True),
    StructField("city", StringType(), True),
    StructField("created_date", StringType(), True),
    StructField("modified_date", StringType(), True)
])

# Create DataFrame with same schema
suppliers_incremental = spark.createDataFrame(data, schema)

# Convert date strings to timestamps
suppliers_incremental = suppliers_incremental \
    .withColumn("created_date", to_timestamp("created_date", "dd-MM-yyyy")) \
    .withColumn("modified_date", to_timestamp("modified_date", "dd-MM-yyyy"))

# Use Delta merge for upsert (update existing + insert new)
from delta.tables import DeltaTable

delta_table = DeltaTable.forName(spark, "catalog_project1.source1.suppliers_raw")

delta_table.alias("target").merge(
    suppliers_incremental.alias("source"),
    "target.supplier_id = source.supplier_id"
).whenMatchedUpdateAll(
).whenNotMatchedInsertAll(
).execute()

print("Incremental load completed: 1 record updated, 1 record inserted")
```
