```python
from pyspark.sql import Row
from pyspark.sql.types import *
from pyspark.sql.functions import to_timestamp 

data = [
    Row(order_item_id=30001, order_id=20001, product_id=501, quantity=1, unit_price=85000, created_date="01-03-2025", modified_date="01-03-2025"),
    Row(order_item_id=30002, order_id=20001, product_id=502, quantity=1, unit_price=2500, created_date="02-03-2025", modified_date="02-03-2025"),
    Row(order_item_id=30003, order_id=20002, product_id=503, quantity=1, unit_price=6500, created_date="03-03-2025", modified_date="03-03-2025"),
    Row(order_item_id=30004, order_id=20003, product_id=504, quantity=1, unit_price=12000, created_date="04-03-2025", modified_date="04-03-2025"),
    Row(order_item_id=30005, order_id=20003, product_id=506, quantity=2, unit_price=500, created_date="05-03-2025", modified_date="05-03-2025"),
    Row(order_item_id=30006, order_id=20005, product_id=510, quantity=1, unit_price=22000, created_date="01-03-2025", modified_date="01-03-2025"),
    Row(order_item_id=30007, order_id=20006, product_id=507, quantity=2, unit_price=3500, created_date="02-03-2025", modified_date="02-03-2025"),
    Row(order_item_id=30008, order_id=20007, product_id=505, quantity=1, unit_price=15000, created_date="03-03-2025", modified_date="03-03-2025"),
    Row(order_item_id=30009, order_id=20009, product_id=509, quantity=2, unit_price=2200, created_date="04-03-2025", modified_date="04-03-2025"),
    Row(order_item_id=30010, order_id=20010, product_id=508, quantity=1, unit_price=5500, created_date="05-03-2025", modified_date="05-03-2025")
]

schema = StructType([
    StructField("order_item_id", IntegerType(), False),
    StructField("order_id", IntegerType(), False),
    StructField("product_id", IntegerType(), False),
    StructField("quantity", IntegerType(), False),
    StructField("unit_price", IntegerType(), False),
    StructField("created_date", StringType(), False),
    StructField("modified_date", StringType(), False)
])

order_items_raw = spark.createDataFrame(data, schema)

order_items_raw = order_items_raw \
    .withColumn("created_date", to_timestamp("created_date", "dd-MM-yyyy")) \
    .withColumn("modified_date", to_timestamp("modified_date", "dd-MM-yyyy"))

display(order_items_raw)
```

```python
# spark.sql("DROP TABLE IF EXISTS catalog_project1.source1.order_items_raw")
order_items_raw.write.mode("append")\
                  .format("delta")\
                  .option("mergeSchema", "true")\
                  .saveAsTable("catalog_project1.source1.order_items_raw")
```

```python
spark.sql("""
SELECT * 
FROM catalog_project1.source1.order_items_raw 
WHERE to_date(modified_date, 'dd-MM-yyyy') < to_date(created_date, 'dd-MM-yyyy')
""").display()
```
